```python
def format_one_line(text: str) -> str:
    """
    Takes a text input and:
    - Removes all newlines
    - Collapses extra spaces
    - Keeps normal capitalization
    - Wraps the result in triple backticks (code block style)
    """
    one_line = " ".join(text.split())
    return f"\n{one_line}\n"


# Example usage
input_text = """
Part 9. General Eligibility and Inadmissibility Grounds – Miscellaneous Conduct: If you answered "Yes" to Item Number 85., what was your nationality or immigration status immediately before you left (for 
example, U.S. citizen or national, lawful permanent resident, nonimmigrant, parolee, present without admission or parole, or any 
other status)? Box (safe max 30  letters)
"""

print(format_one_line(input_text))
```
