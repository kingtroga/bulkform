```python
def format_one_line(text: str) -> str:
    """
    Takes a text input and:
    - Removes all newlines
    - Collapses extra spaces
    - Keeps normal capitalization
    - Wraps the result in triple backticks (code block style)
    """
    one_line = " ".join(text.split())
    return f"\n{one_line}\n"


# Example usage
input_text = """
Part 13. Signature at Interview:   I swear (affirm) and certify under penalty of perjury under the laws of the United States of America that I know that the contents of I swear (affirm) and certify under penalty of perjury under the laws of the United States of America that I know that the contents of [BOX 1] changes made to this application, numbered [BOX 2]  through [BOX 3] , are complete, true and correct. All information on additional pages submitted by me with this Form I-485, on numbered pages [BOX 4] through [BOX 5] are complete, true, and correct.  All documents submitted at this interview were provided by me and are complete, true, and correct. Subscribed to and sworn to (affirmed) before me (Box 1 safe 10 letters)
"""

print(format_one_line(input_text))
```
