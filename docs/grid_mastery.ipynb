Let's start with something you already know: finding locations.

If I say "meet me at the corner of 5th Avenue and 42nd Street",
you know EXACTLY where to go because we have:
- A horizontal reference (5th Avenue) - like X
- A vertical reference (42nd Street) - like Y

PDF coordinates work the same way!

```python
x_position = 100  # 100 points from the LEFT edge
y_position = 200  # 200 points from the BOTTOM edge (in PDFs)

print("=" * 60)
print("UNDERSTANDING COORDINATES")
print("=" * 60)
print(f"\nIf we place text at point ({x_position}, {y_position}):")
print(f"  - It will be {x_position} points from the LEFT edge")
print(f"  - It will be {y_position} points from the BOTTOM edge")
print("\nRemember: In PDFs, (0,0) is at the BOTTOM-LEFT corner")
print("(This is different from images where (0,0) is at TOP-LEFT)")
```

Before we can grid anything, we need to know HOW BIG our canvas is.
Think of it like: "How big is the tablecloth before we draw grid lines?"

```python
from reportlab.lib.pagesizes import A4

# A4 is a standard paper size used almost everywhere except USA
# (USA uses "Letter" size, but A4 is more common globally)
width, height = A4

print("\n" + "=" * 60)
print("A4 PAPER DIMENSIONS")
print("=" * 60)
print(f"\nWidth:  {width:.2f} points")
print(f"Height: {height:.2f} points")

# Let's convert to more familiar units
# 1 point = 1/72 inch (this is a standard printing measurement)
width_inches = width / 72
height_inches = height / 72

print(f"\nIn inches:")
print(f"Width:  {width_inches:.2f} inches")
print(f"Height: {height_inches:.2f} inches")

# And in millimeters (more common internationally)
width_mm = width_inches * 25.4
height_mm = height_inches * 25.4

print(f"\nIn millimeters:")
print(f"Width:  {width_mm:.0f} mm")
print(f"Height: {height_mm:.0f} mm")

print("\nA4 is approximately 8.3 × 11.7 inches (210 × 297 mm)")
```

Now the KEY CONCEPT: What if we divide our A4 paper into equal parts?

Imagine taking a ruler and drawing lines to create a grid.
Let's say we want a 10×10 grid (10 rows, 10 columns).

```python
grid_size = 10  # Try changing this to 20, 50, or 100 later!

print("\n" + "=" * 60)
print(f"DIVIDING A4 INTO A {grid_size}×{grid_size} GRID")
print("=" * 60)

# Calculate how big each cell (square) will be
cell_width = width / grid_size
cell_height = height / grid_size

print(f"\nOur A4 paper is {width:.2f} points wide")
print(f"We want to divide it into {grid_size} columns")
print(f"So each cell will be: {width:.2f} ÷ {grid_size} = {cell_width:.2f} points wide")

print(f"\nOur A4 paper is {height:.2f} points tall")
print(f"We want to divide it into {grid_size} rows")
print(f"So each cell will be: {height:.2f} ÷ {grid_size} = {cell_height:.2f} points tall")

print(f"\n📐 Result: Each grid cell is {cell_width:.2f} × {cell_height:.2f} points")
print(f"   That's about {cell_width/72:.2f} × {cell_height/72:.2f} inches per cell")
```

This is THE MOST IMPORTANT CONCEPT in our entire system.

If a user clicks on grid square (5, 7), how do we know where that is
in actual PDF points so we can place text there?

THE FORMULA:
  pdf_x = grid_x × cell_width
  pdf_y = grid_y × cell_height

Let's see it in action:

```python
print("\n" + "=" * 60)
print("THE MAGIC FORMULA: GRID → PDF COORDINATES")
print("=" * 60)

# Let's say the user wants to place text in grid square (5, 7)
grid_x = 5
grid_y = 7

# Convert to PDF points using our formula
pdf_x = grid_x * cell_width
pdf_y = grid_y * cell_height

print(f"\nUser clicks on grid square ({grid_x}, {grid_y})")
print(f"\nCalculation:")
print(f"  pdf_x = {grid_x} × {cell_width:.2f} = {pdf_x:.2f} points")
print(f"  pdf_y = {grid_y} × {cell_height:.2f} = {pdf_y:.2f} points")
print(f"\n✅ So we place text at PDF coordinate ({pdf_x:.2f}, {pdf_y:.2f})")

# Let's try a few more examples
print("\n" + "-" * 60)
print("MORE EXAMPLES:")
print("-" * 60)

test_cases = [(0, 0), (5, 5), (9, 9)]  # Try different grid positions

for gx, gy in test_cases:
    px = gx * cell_width
    py = gy * cell_height
    print(f"Grid ({gx}, {gy}) → PDF ({px:.2f}, {py:.2f}) points")
```

Let's visualize why this formula works with a simple example.

Imagine a 3×3 grid on a 300×300 point paper:


```python
print("\n" + "=" * 60)
print("VISUAL UNDERSTANDING: 3×3 Grid Example")
print("=" * 60)

simple_width = 300
simple_height = 300
simple_grid = 3

simple_cell_width = simple_width / simple_grid  # 300 ÷ 3 = 100
simple_cell_height = simple_height / simple_grid  # 300 ÷ 3 = 100

print(f"\nPaper: {simple_width}×{simple_height} points")
print(f"Grid: {simple_grid}×{simple_grid}")
print(f"Cell size: {simple_cell_width}×{simple_cell_height} points")

print("\n📊 Grid visualization:")
print("     0        1        2        3")
print("     |--------|--------|--------|")
print("  3  |  (0,2) |  (1,2) |  (2,2) |")
print("     |--------|--------|--------|")
print("  2  |  (0,1) |  (1,1) |  (2,1) |")
print("     |--------|--------|--------|")
print("  1  |  (0,0) |  (1,0) |  (2,0) |")
print("     |--------|--------|--------|")
print("  0")

print("\nIf we want grid cell (1, 1) - the CENTER cell:")
print(f"  pdf_x = 1 × {simple_cell_width} = {1 * simple_cell_width} points")
print(f"  pdf_y = 1 × {simple_cell_height} = {1 * simple_cell_height} points")
print("\nSo the text starts at (100, 100) - exactly at the start of the middle cell!")
```

GRID MASTERY - LESSON 2: Drawing a Grid on a PDF
================================================

In Lesson 1, you learned the theory. Now we'll make it REAL.
We're going to draw actual lines on a PDF to create a visible grid.

Run each cell and observe the output files!


```python
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas
```

Before we draw grids, let's understand how to create a PDF.

Think of it like getting a blank piece of paper and a pen.
The 'canvas' is your blank paper.

```python

print("=" * 60)
print("CREATING YOUR FIRST PDF")
print("=" * 60)

# Step 1: Create a canvas (blank paper)
c = canvas.Canvas("my_first_pdf.pdf", pagesize=A4)

# Get the dimensions
width, height = A4
print(f"\n✅ Created a blank PDF canvas")
print(f"   Size: {width:.2f} × {height:.2f} points (A4)")

# Step 2: Draw something simple - a single line
# A line needs: starting point (x1, y1) and ending point (x2, y2)
x1, y1 = 100, 100  # Start at (100, 100)
x2, y2 = 400, 100  # End at (400, 100) - horizontal line

c.line(x1, y1, x2, y2)
print(f"\n✍️  Drew a horizontal line from ({x1}, {y1}) to ({x2}, {y2})")

# Step 3: Add some text so we can see where things are
c.drawString(100, 110, "This is my first line!")
print(f"   Added text above the line")

# Step 4: IMPORTANT - Save the PDF (like pressing "Save" button)
c.save()
print(f"\n💾 Saved as: my_first_pdf.pdf")
print("   Open this file to see your creation!")
```

Now let's draw a simple grid. We'll start small (3×3) so you can
see each line clearly and understand what's happening.

A grid needs:
- Vertical lines (going up and down)
- Horizontal lines (going left and right)

```python
print("\n" + "=" * 60)
print("DRAWING A 3×3 GRID")
print("=" * 60)

# Create new PDF
c = canvas.Canvas("simple_3x3_grid.pdf", pagesize=A4)
width, height = A4

# Define our grid
grid_size = 3
cell_width = width / grid_size
cell_height = height / grid_size

print(f"\nGrid setup:")
print(f"  Grid size: {grid_size}×{grid_size}")
print(f"  Cell size: {cell_width:.2f} × {cell_height:.2f} points")

# VERTICAL LINES (up and down)
print(f"\n📏 Drawing {grid_size + 1} vertical lines...")
for i in range(grid_size + 1):  # 0, 1, 2, 3 (that's 4 lines for 3 cells)
    # Calculate x position for this vertical line
    x = i * cell_width
    
    # Draw line from bottom (y=0) to top (y=height)
    c.line(x, 0, x, height)
    print(f"   Line {i}: x={x:.2f} (from bottom to top)")

# HORIZONTAL LINES (left to right)
print(f"\n📏 Drawing {grid_size + 1} horizontal lines...")
for i in range(grid_size + 1):  # 0, 1, 2, 3 (that's 4 lines for 3 cells)
    # Calculate y position for this horizontal line
    y = i * cell_height
    
    # Draw line from left (x=0) to right (x=width)
    c.line(0, y, width, y)
    print(f"   Line {i}: y={y:.2f} (from left to right)")

c.save()
print(f"\n✅ Saved as: simple_3x3_grid.pdf")
print("   Open it and count: you should see 4 vertical and 4 horizontal lines")
print("   Creating a 3×3 grid of cells (9 squares total)")
```
